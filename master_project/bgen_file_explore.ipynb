```python
import bgen_reader
```

```python
bgen_file = bgen_reader.open_bgen("data/dadandchicks_indexed_1.bgen")
```

```python
print("number of individuals", bgen_file.nsamples)
```

```python
print("snp-s number", bgen_file.nvariants)
```

```python
print("First 5 variant IDs:", bgen_file.rsids[:5])
```

```python
print("\nFirst 5 variants:")
for i in range(5):
    print(f"Variant {i+1}:")
    print("  rsid      :", bgen_file.rsids[i])
    print("  chromosome:", bgen_file.chromosomes[i])
    print("  position  :", bgen_file.positions[i])
    print("  alleles   :", bgen_file.allele_ids[i])
    print("-" * 40)
```
