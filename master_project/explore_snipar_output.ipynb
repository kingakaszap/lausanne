```python
import h5py as hp
import numpy as np
import pandas as pd
```

```python
with hp.File("data/Super-Scaffold_3_no_mum.hdf5", "r") as output:
    print("Top-level keys:", list(output.keys()))
    for key in output.keys():
        obj = output[key]
        if isinstance(obj, hp.Dataset):
            print(f"{key} -> dataset, shape={obj.shape}, dtype={obj.dtype}")
        else:
            print(f"{key} -> group")
```

```python
import h5py
with h5py.File("data/Super-Scaffold_3_no_mum.hdf5", "r") as f:
    print("bim_columns:", [c.decode() for c in f["bim_columns"][:]])
    print("bim_values:", [v.decode() if isinstance(v, bytes) else v for v in f["bim_values"][:][0]])
    print("imputed_par_gts shape:", f["imputed_par_gts"].shape)
    print("families:", [x.decode() for x in f["families"][:10]])
```
